```python

# Sample DataFrame
data = {
    'A': [1, 2, 2, 2, 5],
    'B': [5, 4, 3, 2, 1]
}
df = pd.DataFrame(data)

value_to_check = 2
n = 2

def check_previous_n_rows(row_index, n, value_to_check, df):
    
    if row_index < n:
        # If we're near the start of the DataFrame, adjust the range
        start_index = 0
    else:
        start_index = row_index - n
    
    # Count occurrences of value_to_check in the previous n rows
    count = (df.iloc[start_index:row_index]['A'] == value_to_check).count()
    
    return count >= 3

# Apply the function to the DataFrame
df['result'] = df['A'].index.map(lambda x: check_previous_n_rows(x, n, value_to_check, df))
df
```
