```python
import pandas as pd
import numpy as np
import pandas_ta as ta
from datetime import datetime, timezone
import plotly.io as pio
import plotly.graph_objects as go  
from sklearn.datasets import load_iris

store=pd.HDFStore("long_backtest_data.h5", "r")  
df = store.select('/data')
df.reset_index(drop=True, inplace=True)
df.notnull().all(axis=1)
store.close() 

#result = df[0:100]
#result = df[df["symbol"]=='MSFT' and df["gain or loss"]<0]

#result = df.loc[(df["symbol"]=='AAPL') & (df["gain_or_loss"]<0)]
#result = result.sort_values("gain_or_loss", ascending=True)
#result = df['symbol'].unique()

result = round(df.groupby('symbol').agg(balance=('balance', 'max')),2)
result['gains'] = round(result-10000, 2)
result['percentage'] =  round((result['gains']/10000) * 100, 2)
result["balance"] = result["balance"].map("${:,.0f}".format)

df_summary = result.groupby(['symbol','balance','percentage'])['gains'].sum().sort_values(ascending=False)
df_summary.loc['Grand Total'] = df_summary.sum()
df_summary = df_summary.reset_index()

df_summary['gains'] = df_summary['gains'].map("${:,.0f}".format)

pd.set_option("display.max_columns", None) 
pd.set_option("display.max_rows", None) 
display(df_summary) 
```
